```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
# Read the dataset
df = pd.read_csv('height-weight.csv')
df.head()
```

```python
df.shape
```

```python
plt.scatter(df['Weight'], df['Height'])
plt.xlabel('Weight')
plt.ylabel('Height')
plt.title('Weight vs Height')
plt.show()
```

```python
df
```

```python
## Divide the dataset into independent and dependent variables
X = df[['Weight']] # Independent feature
y = df['Height'] # Dependent feature
```

```python
X, y
```

```python
# Train Test Split
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.30, random_state = 0)
```

```python
X_train
```

```python
X.shape, y.shape
```

```python
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
# ## Standardize the dataset over independent train data
# from sklearn.preprocessing import StandardScaler
# scaler = StandardScaler()
```

```python
# X_train = scaler.fit_transform(X_train)
# X_test = scaler.transform(X_test)
```

```python
plt.scatter(X_train, y_train)
```

```python
# Train the model Simple Linear Regression
from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

```python
model
```

```python
model.fit(X_train, y_train)
```

```python
print("The slope or coefficient of weight is: ", model.coef_)
print("The intercept of the line is: ", model.intercept_)
```

```python
plt.scatter(X_train, y_train)
plt.plot(X_train, model.predict(X_train), 'g')
plt.show()
```

```python
plt.scatter(X_test,y_test)
plt.plot(X_test, model.predict(X_test), 'r')
plt.show()
```

```python
y_pred_test = model.predict(X_test)
```

```python
y_pred_test, y_test
```

```python
plt.scatter(X_test, y_test)
plt.scatter(X_test, model.predict(X_test), c= 'r')
plt.show()
```

Performance Metrics

 MSE, RMSE, MAE

  R2 Score, Adjusted R2 Score

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error
```

```python
mse = mean_squared_error(y_test, y_pred_test)
mae = mean_absolute_error(y_test, y_pred_test)
rmse = np.sqrt(mse)
print("Mean Squared Error: ", mse)
print("Mean Absolute Error: ", mae)
print("Root Mean Squared Error: ", rmse)
```

```python
from sklearn.metrics import r2_score
score = r2_score(y_test, y_pred_test)
print(score)
print("R2 Score: ", score * 100, '%')
```

```python
X_test.shape
```

```python
X_test.shape[1]
```

```python
# Display Adjusted R2 Score
adjusted_r2 = 1 - (1 - score) * (len(y_test) - 1) / (len(y_test) - X_test.shape[1] - 1)
print("Adjusted R2 Score: ", adjusted_r2 * 100, '%')
```
